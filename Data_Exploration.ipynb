## 1. Import Required Libraries

In this section, we import Python libraries required for:
- Data manipulation
- Numerical operations
- Data visualization
- Dataset loading

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris


```

## 2. Loading Iris Dataset

The Iris dataset is a built-in dataset provided by Scikit-Learn.

It contains 150 flower samples with four numerical features.

```python
iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df['Species'] = iris.target_names[iris.target]

df.head() 
```

## 3. Understanding Dataset Structure

We will explore:
- Number of records
- Feature names
- Data types
- Statistical information

```python
df.shape
```

```python
df.info()
```

## 4. Missing Value Analysis

Checking missing values is an important step before performing analysis.

Missing values can affect visualization and machine learning models.

```python
df.isnull().sum()
```

No missing values are present in the dataset.
Therefore, no data cleaning operation is required.

```python
#. Duplicate Records Check
df.duplicated().sum()
```

```python
# Remove duplicate
df.drop_duplicates(inplace=True)
```

## 5. Feature Distribution Analysis

Histograms help us understand how values are distributed across different features.

```python
features = df.columns[:-1]

plt.figure(figsize=(15,10))

for i, feature in enumerate(features):
    plt.subplot(2,2,i+1)
    sns.histplot(
        df[feature],
        kde=True
    )
    plt.title(feature)

plt.tight_layout()
plt.show()
```

## 6. Outlier Detection Using Box Plot

Box plots help identify:
- Data spread
- Median values
- Possible outliers

```python
plt.figure(figsize=(10,5))

sns.boxplot(data=df)

plt.title("Feature Distribution")

plt.xticks(rotation=45)

plt.show()
```

## 7. Species Distribution

Checking the number of samples available for each flower species.

```python
sns.countplot(
    x='Species',
    data=df
)

plt.title("Number of Samples per Species")

plt.show()
```

## 8. Relationship Between Features

Pair plots show relationships between numerical features and help identify patterns among species.

```python
sns.pairplot(
    df,
    hue="Species"
)

plt.show()
```

## 09. Feature Correlation Analysis

Correlation shows the relationship strength between numerical features.

```python
plt.figure(figsize=(8,5))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```
